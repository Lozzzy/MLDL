## 라이브러리 - numpy broadcast  

=> 행렬의 사칙연산은 기본적으로 두 개의 행렬 크기가 같은 경우에만 수행할 수 있음. 그러나 numpy에서는 크기가 다른 두 행렬간에도 사칙연산을 할 수 있는데 이를 브로드 캐스트라고 지칭함  

=> 차원이 작은 쪽인 큰 쪽의 행 단위로 반복적으로 크기를 맞춘 후에 계산

행 또는 열(또는 다차원 배열의 각 축)중 하나이상이 같고, 같지 않을경우 1이어야만 그 축을 따라서 자동으로 확장(복제)되어서 브로드캐스팅이 됩니다.  

사칙연산중 *은 행렬의 곱이 아님 원소별 대응되는 곱을하는것 (2x2) * (2x3) 은 불가능, (2x3) * (2x3) 은 가능. 즉, 행렬의 곱과는 다름

```python
A = np.array([ [1, 2], [3, 4] ])
b = 5

# b 는 [[5, 5], [5, 5]] 가 된다
print(A+b)
```

```python
C = np.array([ [1, 2], [3, 4] ])
D = np.array([4, 5])

# 행 또는 열(또는 다차원 배열의 각 축)중 하나 이상이 서로 같고, 같지 않을경우 1이어야만 그 축을 따라서 자동으로 확장(복제)되어서 브로드캐스팅이 됩니다.
# D 는 [[4, 5], [4, 5]] 가 된다
print(C+D)
```

```python
C = np.array([ [1, 2, 3], [3, 4, 5] ]) # ( 2 x 3 )
D = np.array([[4, 5], [6, 7]]) # ( 2 x 2 )

# 원소의 곱이므로 오류
C*D # (2x3) * (2x2)
D*C # (2x2) * (2x3) 이지만 원소의 갯수가 안맞아 불가능
# 행렬의 곱 은 가능
np.dot(C,D)
np.dot(D,C)
```

```python
C = np.array([ [1, 2, 3], [3, 4, 5] ]) # ( 2 x 3 )
D = np.array([[4, 5], [6, 7], [8,9]]) # ( 3 x 2 )

#원소별 곱이라 오류
C*D
```

```python
C = np.array([ [1, 2, 3], [3, 4, 5] ]) # ( 2 x 3 )
D = np.array([[4, 5, 6], [6, 7, 8]]) # ( 2 x 3 )

C*D
```

## 라이브러리 - 전치행렬  

어떤 행렬의 전치행렬은 원본 행렬의 열은 행으로, 행은 열로 바꾼것으로서, 원본 행렬을 A 라고 하면 전치행렬은 A^t 로 나타냄.  

즉 1행은 1열로 바꾸고 2행은 2열로, 3행은 3열로 바꾼 행렬을 의미

```python
A = np.array([[1,2], [3,4], [5,6]])

B = A.T

print("A.shape == ", A.shape, "B.shape == ", B.shape)
print(A)
print(B)
```

```python
C = np.array([1,2,3,4,5])
D = C.T # vector는 전치행렬(transpose)이 안된다

E = C.reshape(1,5) # 행렬로 변환
F= E.T # 행렬이므로 전치행렬 적용

print("C.shape == ", C.shape, "D.shape == ", D.shape)
print("E.shape == ", E.shape, "F.shape == ", F.shape)
print(C)
print(D)
print(E)
print(F)
```

## 라이브러리 - 행렬 iterator  

명시적 인덱싱 / 슬라이싱 이외에, 행렬 모든 원소를 access하는 경우에는 iterator  사용가능  

https://rphabet.github.io/posts/python_numpy_loop/

```python
import numpy as np

A = np.array([[10, 20, 30, 40], [50, 60, 70, 80]])

print(A, "\n")
print("A.shape == ", A.shape, "\n")

# 행렬 A의 iterator 생성

it = np.nditer(A, flags=['multi_index'], op_flags=['readwrite'])

while not it.finished:
    
    idx = it.multi_index
    print("it.multi_index == ", idx)
    
    print("current value => ", A[idx])
    
    it.iternext()
```

```python
import numpy as np

#array 형태이어야함. B = [1,2,3] 이면 안됨
B = np.array([1,3,4])

print(B, "\n")
print("B.shape == ", B.shape, "\n")

# 벡터 B의 iterator 생성
# C_index는 1차원 배열 즉 벡터 등에서만 사용됨
# 2차원 즉 행렬에는 multi_index가 사용됨
# 앞의 c는 프로그래밍 언어인 c이다. 뜻은 c언어에서 사용하는 index 형태로 iterator를 만들겠다.
# 이 index의 형태가 c언어에서 사용하는 index의 형태와 같다!
# c언어에서 사용하는 인덱스의 형태가 뭔데? ==> 0, 1, 2, 3, 4, .... 우리가 사용하는것과 같음.
it = np.nditer(B, flags=["c_index"])

while not it.finished:
    
    idx = it.index
    print("it.index == ", it.index)
    
    print("current value => ", B[idx])
    
    it.iternext()
```

```python
import numpy as np

#array 형태이어야함. B = [1,2,3] 이면 안됨
B = [1,3,4]

print(B, "\n")

it = np.nditer(B, flags=["c_index"])

while not it.finished:
    
    idx = it.index
    print("it.index == ", it.index)
    
    print("current value => ", B[idx])
    
    it.iternext()
```

## 라이브러리 - concatenate

행렬에 행 또는 열 추가하기 위한 라이브러리  

머신러닝의 회귀 코드 구현시 가중치와 바이어스를 별도로 구분하지 않고 하나의 행렬로 취급하기 위한 프로그래밍 구현 기술

```python
A = np.array([[10, 20, 30], [40, 50, 60]])

print(A.shape)

print(np.array([70, 80, 90]).shape)
# 행을 추가하기 위해선 기존 행렬에서 열에 맞게 reshape
# vector는 행과 열을 구분하지 않기때문에 2차원 행렬로 변경
row_add = np.array([70, 80, 90]).reshape(1, 3)
print(row_add.shape)

# 열을 추가하기 위해선 기존 행렬에서 행에 맞게 reshape
column_add = np.array([1000, 2000]).reshape(2, 1)
print(column_add.shape)

# axis = 0 은 행추가
B = np.concatenate((A, row_add), axis=0)

print(B)

# axis = 1 은 열추가
C = np.concatenate((A, column_add), axis=1)

print(C)
```

```python
# 무조건 포물선 형태로 되도록 그래프를 그리기
import matplotlib.pyplot as plt

%matplotlib inline

x_data = np.random.rand(10)*10
for a in range(9):
    if x_data[a+1] < x_data[a]:
        x_data[a+1] += x_data[a]

y_data = np.random.rand(10)*10
for i in range(9):
    for c in range(9):
        if y_data[c] == y_data[c+1]:
            y_data[c+1] += 1
    if i < 5:
        for a in range(5):
            for b in range(5):
                if y_data[a] > y_data[b]:
                    z = y_data[a]
                    y_data[a] = y_data[b]
                    y_data[b] = z
        
    elif i > 4:
        for a in range(5):
            for b in range(5):
                if y_data[a+5] < y_data[b+5]:
                    z = y_data[a+5]
                    y_data[a+5] = y_data[b+5]
                    y_data[b+5] = z
        
                    
plt.title('line plot')
plt.grid()
plt.plot(x_data, y_data, color='b')
plt.show()
```

```python
# 개선버전
import matplotlib.pyplot as plt
import numpy as np

x_data = np.random.rand(10)*10
for i in range(9):
    if x_data[i+1] < x_data[i]:
        x_data[i+1] += x_data[i]

y_data = np.random.rand(10)*10

# 중복 제거(인접값 같으면 뒤에 +1)
for _ in range(9):
    for c in range(9):
        if y_data[c] == y_data[c+1]:
            y_data[c+1] += 1

# 앞 절반 내림차순, 뒤 절반 오름차순
for _ in range(5):
    for a in range(5):
        for b in range(5):
            if y_data[a] > y_data[b]:
                y_data[a], y_data[b] = y_data[b], y_data[a]
            if y_data[a+5] < y_data[b+5]:
                y_data[a+5], y_data[b+5] = y_data[b+5], y_data[a+5]

plt.title('line plot')
plt.grid()
plt.plot(x_data, y_data, color='b')
plt.show()
```
