# list comprehension

리스트의 [...] 괄호 안에 for 루프를 사용하여 반복적으로 표현식(expression)을 실행해서 리스트 요소들을 정의 하는 방법을 List Comprehension이라 하며, 머신러닝 코드에서 자주 사용되는 기법

```python
list_data = [ x**2 for x in range(5) ]

print(list_data)
```

```python
raw_data = [ [1,10], [2,15], [3,30], [4,55] ]

all_data = [ x for x in raw_data ]
x_data = [ x[0] for x in raw_data ]
y_data = [ x[1] for x in raw_data ]

print("all_data == ", all_data)
print("x_data == ", x_data)
print("y_data == ", y_data)
```

default parameter

=> 함수의 입력 파라미터에 기본 값을 지정하는 것을 말하며, 이러한 디폴트 파라미터는, 함수가 호출되었을 경우 입력파라미터에 명시적인 값이 전달되지 않으면 기본으로 지정한 값을 사용하곘다는 의미

```python
def print_name(name, count=2):
    
    for i in range(count):
        print("name ==", name)
    

print_name("DAVE")
```

mutable / immutable parameter

=> 입력 파라미터가 mutable(list, dict, numpy 등) 데이터형인 경우는 원래의 데이터에 변형이 일어남  
=> immutable(숫자, 문자, tuple 등)은 원래의 데이터에 변형이 일어나지 않음

```python
def mutable_immutable_func(int_x, input_list):
    
    int_x += 1
    input_list.append(100)

x = 1
test_list = [ 1, 2, 3 ]

mutable_immutable_func(x, test_list)

print("x==", x, ", test_list == ", test_list)
```

# 함수 - lambda

람다는 함 줄로 함수를 작성하는 방법으로, 익명함수 또는 람다표현식 등으로 불림  
  
lambda는 다른 함수의 인수(parameter)로 넣을때 주로 사용하며, 특히 머신러닝에서 미분을 계산하기 위해 필요한 수치미분과 활성화 함수등을 표현할 때, 수학에서 함수 표현처럼 f(x), f(x,y)등으로 사용되며, 마지막 표현으로 대체된다는 의미

```python
f = lambda x : x+100 # return (x = x+100)

for i in range(3):
    print(f(i))
```

# 클래스 - class

```python
class Person:
    def __init__(self, name):
        self.name = name
        print(self.name + " is initialized")
        
    def work(self, company):
        print(self.name + " is working in " + company)
        
    def sleep(self):
        print(self.name + "is sleeping")
        
obj = Person("PARK")

obj.work("ABCDEF")
Person.work(obj, "ABCDEF")

obj.sleep()


print("current person object is ", obj.name)
```

클래스 변수  
=> 클래스 변수는 클래스 내외부에서 "클래스명.클래스 변수명" 으로 접근 할 수 있음  
  
클래스 메서드  
는 메서드 앞에 @classmethod를 반드시 표시하여 해당 메서드가 클래스 메서드임을 표시함.  
=> 클래스 메서드는 객체 인스턴스를 의미하는 self 대신 cls 라는 클래스를 의미하는 파라미터를 인수로 전달받음


```python
class Person:
    
    count = 0 
    
    def __init__(self, name):
        self.name = name
        self.count += 1
        Person.count += 2
        print(self.name + " is initialized")
        
    def work(self, company):
        print(self.name + "is working in" + company)
        
    def sleep(self):
        print(self.name + " is sleeping")
        
    @classmethod
    def getCount(cls):
        print("This is getCount Method")
        return cls.count

obj1 = Person("PARK")
obj2 = Person("KIM")

obj1.work("Google")

obj2.sleep()

print("current person object is ", obj1.name, ", ", obj2.name)

print("Person getCount == ", Person.getCount())

# 파이썬 객체에서 self.count처럼 어떤 속성에 접근하려고 하면, Python은 아래 순서로 찾습니다:

# 1. 인스턴스 내부(obj1, obj2)의 변수

# 2. 없으면 → 클래스의 변수 (Person.count)

# 3. 없으면 → 에러 (AttributeError)
# obj.count 즉 self.count 변수 선언이 안되어있기 때문에 Person.count를 참조하여 self.count 를 만듬 즉 self.count = Person.count 로 선언 후 self.count +=1

# __init__에서 self.count += 1 은
# self.count가 없기 때문에 → Person.count를 참조해 계산하고
# 그 결과로 새로운 인스턴스 변수 self.count를 생성함

# 따라서 obj2.count 일때에는 Person.count가 2이기때문에 3이 된다  만약 클래스에 self.count = 0 이라는 변수 선언이 있었다면 둘다 1이 된다.
print("obj1 count == ", obj1.count)
print("obj2 count == ", obj2.count)
#obj 로 호출을 하든 Person으로 호촐을 하든 return 값은 cls.count. 즉 클래스 변수값이 리턴됨. 이 둘의 차이는 인스턴스호출과 클래스호출의 차이 즉 이 호출이 클래스 레벨인지 인스턴스 레벨인지 확인용도정도.
print("obj1 getCount == ", obj1.getCount())

# Person.count와 Person.getCount 의 차이점은 메서드내에 다른 로직이 추가될수 있음 여기선 print("This is getCount Method")임
print("Person count == ", Person.count)
```

Private 변수와 메서드

```python
class PrivateMemberTest:
    
    def __init__(self, name1, name2):
        
        self.name1 = name1
        #private 변수 선언
        self.__name2 = name2
        print("initialized with " + name1 + "," + name2)
        
    def getNames(self):
        #클래스 내에서는 private 메서드 또는 변수에 접근 가능
        self.__printNames()
        return self.name1, self.__name2
    
    def __printNames(self):
        print(self.name1, self.__name2)
        
obj = PrivateMemberTest("PARK", "KIM")

print(obj.name1)
print(obj.getNames())
#클래스 밖에서는 private 메서드 혹은 변수에 접근이 불가능 하기에 에러발생
print(obj.__printNames())
print(obj.__name2)
```

외부함수와 클래스 메서드이름이 같은 경우  
=> c++, Java 언어의 this 처럼 self를 통해 method 호출  
=> self 를 붙이지 않으면 동일한 이름의 외부 함수 호출됨

```python
def print_name(name):
    print("[def] ", name)
    
class SameTest:
    def __init__(self):
        pass
    
    def print_name(self, name):
        print("[SameTest] ", name)
        
    def call_test(self):
        print_name("KIM")
        self.print_name("KIM")
        
obj = SameTest()

print_name("LEE")
obj.print_name("LEE")
obj.call_test()
```

# 예외처리 - exception


```python
def calc(list_data):
    
    sum = 0
    
    try:
        sum = list_data[0] + list_data[1] + list_data[2]
        
        if sum < 0:
            raise Exception("Sum is Minus")
        
    except IndexError as err:
        print(str(err))
    except Exception as err:
        print(str(err))
    finally:
        print(sum)
```

```python
# index error 발생
calc([1,2])
```

```python
# 인위적인 excetpion 발생
calc([1,3,-100])
```

# with 구문
일반적으로 파일 또는 세션을 사용하는 순서는 다음과 같음  
open() -> read() 또는 write() -> close()  

그러나 파이썬에서 with 구문을 사용하면 명시적으로 리소스 close() 를 해주지 않아도 자동으로 close() 해주는 기능이 있음. with 블록을 벗어나는 순간 파일이나 세션 등의 리소스를 자동으로 close 시킴  

=> 딥러닝 프레임워크인 TensorFlow 의 session 사용시 자주 사용됨

```python
# 일반적인 방법
f = open("./file_test", 'w')

f.write("Hello, Python!!")

f.close()
```

```python
# with 구문을 사용한 방법
# with 블록을 벗어나는 순간 파일 객체 f가 자동으로 close 됨.
with open("./file_test2", 'w') as f:
    f.write("Hello, Python!!")
```
